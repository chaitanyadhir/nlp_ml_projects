Importing all essential libraries

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import re
import warnings
warnings.filterwarnings('ignore')
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

Storing the csv into a variable using pandas.

```python
df=pd.read_csv('IMDB Dataset.csv')
#https://www.kaggle.com/datasets/lakshmi25npathi/imdb-dataset-of-50k-movie-reviews?resource=download
```

Checking the data's information and structure.

Creating a function to clean the data i.e. removing all unnecessary elements that might interfere with the prediction.

```python
def cleaning(text):
    text=text.lower()
    text=re.sub(r'<.*?>','',text)
    text=re.sub(r'[^a-zA-Z\s]','',text)
    tokens=word_tokenize(text)
    stop_words=set(stopwords.words('english'))
    tokens=[token for token in tokens if token not in stop_words]
    return ' '.join(tokens)
```

Calling the cleaning function to create a new column of cleaned review for using as a better parameter.

```python
df['cleaned_review'] = df['review'].apply(cleaning)
```

```python
df.head()
```

Extracting feature using tfid Vectorization.

```python
tfidf = TfidfVectorizer(max_features=5000)
```

Converting text to tdif features.

```python
x=tfidf.fit_transform(df['cleaned_review'])
y=df['sentiment'].map({'positive':1, 'negative':0})
```

Setting up the variables for model to train.

```python
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
print(X_train.shape,y_train.shape)
```

Choosing a model: here we are choosing logisticRegression.

```python
model=LogisticRegression(max_iter=1000)
model.fit(X_train,y_train)
```

Finally predicting. after training the model.

```python
y_pred=model.predict(X_test)
```

Checking the matrixes.

```python
acc=accuracy_score(y_test,y_pred)
prec=precision_score(y_test,y_pred)
recall=recall_score(y_test,y_pred)
```

```python
print('Model prediction accuracy: ',acc*100,'%')
print('Model prediction precision: ',prec*100,'%')
print('Model prediction recall: ',recall*100,'%')
```

Automating the process for tesing on sample input.

```python
def predict(text):
    text2=cleaning(text)
    para=tfidf.transform([text2])
    prediction=model.predict(para)
    return 'postive' if prediction[0]==1 else 'negative'
```

Testing on sample input.

```python
reviews = [
    "This movie was okay to watch but could have been better.",
    "I was unsure about this movie but it was a really good watch.",
]
for i in reviews:
    print(f"\nReview: ",i)
    print(f"Predicted Sentiment: ",predict(i))
```
